### Dependencies

```python
# Import dependencies
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', None)

## Set working directory to tree_growth_metaanalysis_snmc
os.chdir('../../')
```

```python
## Metaregressor file directory
mr_dir = 'Covariate_and_metaregressor_processing/Processed_data/'
```

```python
## List of sites
site_list = ['Blodgett', 'LaTour', 'Sequoia', 'STEF', 'Teakettle', 'TharpsCreek', 'WLakeTahoe']
```

## Climate metaregressors

```python
### Load climate metaregressors processed via create_climate_metaregressors
clim = pd.read_csv('{}Climate/site_mean_clim.csv'.format(mr_dir))
clim
```

## Soil metaregressors

```python
### Load soil metaregressors processed via create_soil_metaregressors
soils = pd.read_csv('{}Soil/unit_mean_soil.csv'.format(mr_dir))

## Make Teakettle unit names lowercase
soils.loc[soils['Site']=='Teakettle', 'UnitID'] = soils.loc[soils['Site']=='Teakettle', 'UnitID'].str.lower()

soils
```

## Relative stand density index

```python
## Compile rSDI for all sites
rsdi = pd.DataFrame()

for site in site_list:
    temp = pd.read_csv(mr_dir + 'Stand_structure/{}_rsdi.csv'.format(site))
    temp['Site'] = site
    rsdi = pd.concat([rsdi, temp])

## Make UnitID a string
rsdi['UnitID'] = rsdi['UnitID'].astype(str)

rsdi
```

## Treatment intensity

```python
## Compile treatment intensity for all sites
intensity = pd.DataFrame()

for site in site_list:
    temp = pd.read_csv(mr_dir + 'Stand_structure/{}_intensity.csv'.format(site))
    temp['Site'] = site
    intensity = pd.concat([intensity, temp])

## Replace NaN treatment with "None"
intensity.loc[intensity['Treatment'].isnull(), 'Treatment'] = 'None'

## Make UnitID a string
intensity['UnitID'] = intensity['UnitID'].astype(str)

## Replace null intensities with 0
intensity.loc[intensity['BA_reduction'].isnull(), 'BA_reduction'] = 0

intensity
```

# Merge metaregressors

```python
## Merge climate and soil
all_rgrs = pd.merge(soils[['Site', 'UnitID', 'TotalDepth', 'CEC_50cm', 'ksat_50cm', 'AWC_50cm']],
                    clim,
                    how = 'outer',
                    on = 'Site')

## Merge with rSDI
all_rgrs = pd.merge(all_rgrs, 
                    rsdi[['Site', 'UnitID', 'rSDI']],
                   how = 'outer',
                   on = ['Site', 'UnitID'])

## Merge with intensity
all_rgrs = pd.merge(all_rgrs,
                    intensity,
                   how = 'outer',
                   on = ['Site', 'UnitID'])

all_rgrs
```

```python
## Save
all_rgrs.to_csv(mr_dir + 'all_regressors_by_unit.csv',
               index = False)
```
